```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

# Assign colum names to the dataset
names = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'Target']

# Read dataset to pandas dataframe
dataset = pd.read_csv(url, names=names)
```

```python
dataset.head(5)
```

```python
dataset['Target'].unique()
```

```python
dataset.shape
```

```python
X = dataset.iloc[:, 0:4]
Y = dataset.loc[:, "Target"]
```

```python
X.head()
```

```python
Y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size = 0.1, random_state = 77, )
```

```python
X_train.shape
```

```python
linear_SVM = SVC(kernel = "poly", C = 10, gamma = 0.1) # model init
linear_SVM.fit(X_train, y_train) # model training
```

```python
train_predictions = linear_SVM.predict(X_train)
predictions = linear_SVM.predict(X_test)
predictions
```

```python
print(confusion_matrix(y_test, predictions))
print(classification_report(y_test, predictions))
```

```python
print(confusion_matrix(y_train, train_predictions))
print(classification_report(y_train, train_predictions))
```
